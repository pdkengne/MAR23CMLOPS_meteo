```python
# chargement des librairies
import glob as g
import pandas as pd
import numpy as np
from datetime import datetime
import os
```

```python
# noms des colonnes dans les fichiers cvs à manipuler
csv_col_names=['Location', 'Date', 'Minimum temperature (°C)', 'Maximum temperature (°C)',
               'Rainfall (mm)', 'Evaporation (mm)', 'Sunshine (hours)',
               'Direction of maximum wind gust ', 'Speed of maximum wind gust (km/h)',
               'Time of maximum wind gust', '9am Temperature (°C)',
               '9am relative humidity (%)', '9am cloud amount (oktas)',
               '9am wind direction', '9am wind speed (km/h)', '9am MSL pressure (hPa)',
               '3pm Temperature (°C)', '3pm relative humidity (%)',
               '3pm cloud amount (oktas)', '3pm wind direction',
               '3pm wind speed (km/h)', '3pm MSL pressure (hPa)']
```

```python
# chargement du dataset
input_path = "../../datasets/raw/latest/" + "IDCJDW2002.202306.csv"
df = pd.read_csv(input_path, encoding = "ISO-8859-1", names = csv_col_names)
df
```

```python
# extraction du code de la localité
text = df["Location"][1]
location_code=text.split(".")[0].split(" ")[-1]
location_code
```

```python
# ajout du code de la localité dans le dataset
df["Location"] = location_code
df
```

```python
# suppression des lignes ayant au plus 2 valeurs non manquantes
df.dropna(thresh=3, axis=0, inplace=True)
df
```

```python
# suppression de la première ligne
df.drop(index=df.index[0], axis=0, inplace=True)
df
```

```python
# description breve du dataset resultant
df.info()
```

```python
# colonnes des variables numériques
num_cols = ['Minimum temperature (°C)', 'Maximum temperature (°C)', 'Rainfall (mm)', 'Evaporation (mm)', 'Sunshine (hours)',
            'Speed of maximum wind gust (km/h)', '9am Temperature (°C)', '9am relative humidity (%)', '9am cloud amount (oktas)',
            '9am MSL pressure (hPa)', '9am wind speed (km/h)', '3pm Temperature (°C)', '3pm relative humidity (%)', '3pm cloud amount (oktas)',
            '3pm wind speed (km/h)', '3pm MSL pressure (hPa)']

num_cols
```

```python
# changement de la valeur "Calm" par la valeur "1"
df['9am wind speed (km/h)'].replace({"Calm": "1"}, inplace = True)
df['3pm wind speed (km/h)'].replace({"Calm": "1"}, inplace = True)
df
```

```python
#conversion des colonnes de la liste "num_cols" en types numériques
df[num_cols] = df[num_cols].astype(float)
df.info()
```

```python
# conversion de la colonne Date en un object de type datetime
df['Date'] = pd.to_datetime(df['Date'])
# création de nouvelles colonnes reliées à la date
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
```

```python
# extraction d'un nouveau dataset n'incluant pas la colonne "Date"
df_2 = df.drop('Date', axis = 1)
```

```python
df_2.info()
```

```python
# extraction d'un nouveau dataset n'incluant pas la colonne "Time of maximum wind gust"
df_3 = df_2.drop("Time of maximum wind gust", axis = 1)
df_3.info()
```

```python
# creation de la colonne "RainToday"
df_3["RainToday"] = df_3["Rainfall (mm)"] >= 1
# codage de la variable: RainToday
df_3["RainToday"].replace({False:"No", True: "Yes"}, inplace = True)
df_3
```

```python
# creation de la colonne "RainTomorrow"
df_3["RainTomorrow"] = df_3.shift(-1)["RainToday"]
df_3
```
