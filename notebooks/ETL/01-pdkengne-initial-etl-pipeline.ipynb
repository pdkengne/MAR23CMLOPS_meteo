```python
# chargement des librairies
import glob as g
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
# chargement du dataset
input_path = "../../datasets/raw/latest/" + "IDCJDW2002.202306.csv"
df = pd.read_csv(input_path, encoding = "ISO-8859-1", skiprows=7)
# suppression de la première colonne
df.drop(columns=df.columns[0], axis=1, inplace=True)
# affichage du dataset
print(df)
```

```python
num_cols = df.select_dtypes(include = ["int", "float"]).columns
num_cols
```

```python
# description breve du dataset resultant
df.info()
```

```python
# conversion de la colonne Date en un object de type datetime
df['Date'] = pd.to_datetime(df['Date'])
# création de nouvelles colonnes reliées à la date
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
```

```python
# extraction d'un nouveau dataset n'incluant pas la colonne "Date"
df_2 = df.drop('Date', axis = 1)
```

```python
df_2.info()
```

```python
# lecture du code de la localité
text = pd.read_csv(input_path, encoding = "ISO-8859-1", skiprows=1, nrows=1, header=None)
location_code=text.iloc[0,0].split(".")[0].split(" ")[-1]
location_code
```

```python
# ajout du code de la localité dans le nouveau dataset
df_2["Location"] = location_code
df_2.info()
```

```python
# extraction d'un nouveau dataset n'incluant pas la colonne "Time of maximum wind gust"
df_3 = df_2.drop("Time of maximum wind gust", axis = 1)
df_3.info()
```

```python
# creation de la colonne "RainToday"
df_3["RainToday"] = df_3["Rainfall (mm)"] >= 1
# codage de la variable: RainToday
df_3["RainToday"].replace({False:"No", True: "Yes"}, inplace = True)
print(df_3)
```

```python
# creation de la colonne "RainTomorrow"
df_3["RainTomorrow"] = df_3.shift(-1)["RainToday"]
print(df_3)
```
